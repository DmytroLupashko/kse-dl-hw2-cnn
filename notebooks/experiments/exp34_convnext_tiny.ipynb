```python
import matplotlib.pyplot as plt

from core.training.config import ROOT
from core.training.config import VIP
from core.training.cv import CVResult
from core.analysis.gradcam import GradCamFigures
from core.analysis.plots import pr_curves, save_figures, use_style

N_POSITIVES = None  # all labeled positives
CONFIG_DIR = ROOT / "configs" / "models"
CONFIG = CONFIG_DIR / "exp34_convnext_tiny" / "convnext_tiny.yaml"
EXP26 = CONFIG_DIR / "exp26_coarse_dropout_5fold" / "resnet18.yaml"
FIGURES = ROOT / "docs" / "figures"
PREFIX = "exp34_convnext_tiny"

# dataviz reference palette, light mode
SERIES = {"exp26": "#eb6834", "ConvNeXt-Tiny": "#2a78d6"}
use_style()
```

## Training


```python
results = {"exp26": CVResult.load(EXP26), "ConvNeXt-Tiny": CVResult.load(CONFIG)}
```

## Precision-recall curves


```python
fig = pr_curves(results, colors=SERIES, title="Out-of-fold rows of folds 0–4")
figures = {f"{PREFIX}_pr_curves.png": fig}
plt.show()
```

## Grad-CAM VIP: unseen images


```python
cam = GradCamFigures(CONFIG)
for cls in VIP:
    figures[f"{PREFIX}_grad_cam_{cls.lower()}_validation.png"] = cam.positives(
        cls, n=N_POSITIVES
    )
plt.show()
```

```python
save_figures(figures, FIGURES)
```
