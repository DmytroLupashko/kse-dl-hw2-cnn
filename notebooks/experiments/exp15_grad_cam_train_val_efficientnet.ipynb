Card: [exp15_grad_cam_train_val_efficientnet.md](../../docs/neuroslope/experiments/exp15_grad_cam_train_val_efficientnet.md)


```python
import matplotlib.pyplot as plt

from training.config import ROOT
from training.config import VIP
from analysis.gradcam import GradCamFigures
from analysis.plots import save_figures, use_style

N_IMAGES = None  # all labeled positives
CONFIG = ROOT / "configs" / "models" / "exp10_efficientnet" / "efficientnet_b0.yaml"
FIGURES = ROOT / "docs" / "figures"
PREFIX = "exp15_grad_cam_train_val_efficientnet"

use_style()
```

## Fold models


```python
cams = GradCamFigures(CONFIG)
```

## Validation


```python
for cls in VIP:
    fig = cams.positives(cls, train=False, n=N_IMAGES)
    save_figures({f"{PREFIX}_{cls.lower()}_validation.png": fig}, FIGURES, dpi=100)
    plt.show()
```

## Train


```python
for cls in VIP:
    fig = cams.positives(cls, train=True, n=N_IMAGES)
    save_figures({f"{PREFIX}_{cls.lower()}_train.png": fig}, FIGURES, dpi=100)
    plt.show()
```
