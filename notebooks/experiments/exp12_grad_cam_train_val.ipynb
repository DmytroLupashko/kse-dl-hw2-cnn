# Exp 12: Grad-CAM train vs validation

Card: [exp12_grad_cam_train_val.md](../../docs/experiments/exp12_grad_cam_train_val.md)


```python
import matplotlib.pyplot as plt

from core.training.config import ROOT
from core.training.config import VIP
from core.analysis.gradcam import GradCamFigures
from core.analysis.plots import save_figures, use_style

N_IMAGES = None  # all labeled positives
CONFIG = ROOT / "configs" / "models" / "exp11_resnet18_5fold" / "resnet18.yaml"
FIGURES = ROOT / "docs" / "figures"
PREFIX = "exp12_grad_cam_train_val"

use_style()
```

## Fold models


```python
cams = GradCamFigures(CONFIG)
```

## Validation


```python
for cls in VIP:
    fig = cams.positives(cls, train=False, n=N_IMAGES)
    save_figures({f"{PREFIX}_{cls.lower()}_validation.png": fig}, FIGURES, dpi=100)
    plt.show()
```

## Train


```python
for cls in VIP:
    fig = cams.positives(cls, train=True, n=N_IMAGES)
    save_figures({f"{PREFIX}_{cls.lower()}_train.png": fig}, FIGURES, dpi=100)
    plt.show()
```
