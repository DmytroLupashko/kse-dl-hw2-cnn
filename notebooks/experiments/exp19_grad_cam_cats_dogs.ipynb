```python
import os

import matplotlib.pyplot as plt

from core.config import ROOT
from core.cv import Config
from core.gradcam import GradCamFigures

SMOKE = os.environ.get("SMOKE") == "1"
if SMOKE:
    N_IMAGES = 2  # per figure
else:
    N_IMAGES = 16
SEED = 0
CLASSES = ["Cat", "Dog"]
CONFIG = Config.from_toml(
    ROOT / "sections" / "model" / "configs" / "exp16_downsampling" / "resnet18.toml"
)
FIGURES = ROOT / "sections" / "model" / "figures"
PREFIX = "exp19_grad_cam_cats_dogs"
SAVE_FIGURES = not SMOKE

plt.rcParams.update({"figure.facecolor": "#fcfcfb", "savefig.facecolor": "#fcfcfb"})
```

## Fold models


```python
cams = GradCamFigures(CONFIG)
```

## Validation


```python
for cls in CLASSES:
    fig = cams.positives(cls, train=False, n=N_IMAGES, seed=SEED)
    if SAVE_FIGURES:
        fig.savefig(FIGURES / f"{PREFIX}_{cls.lower()}_validation.png", dpi=100)
    plt.show()
```
