```python
import matplotlib.pyplot as plt

from training.config import ROOT

from analysis.gradcam import GradCamFigures
from analysis.plots import save_figures, use_style

N_IMAGES = 16
SEED = 0
CLASSES = ["Cat", "Dog"]
CONFIG = ROOT / "configs" / "models" / "exp16_downsampling" / "resnet18.yaml"
FIGURES = ROOT / "docs" / "figures"
PREFIX = "exp19_grad_cam_cats_dogs"

use_style()
```

## Fold models


```python
cams = GradCamFigures(CONFIG)
```

## Validation


```python
for cls in CLASSES:
    fig = cams.positives(cls, train=False, n=N_IMAGES, seed=SEED)
    save_figures({f"{PREFIX}_{cls.lower()}_validation.png": fig}, FIGURES, dpi=100)
    plt.show()
```
