```python
import pandas as pd

from helpers.config import ROOT, SUBMISSIONS

from helpers.ensemble import MeanEnsemble

CONFIG_DIR = ROOT / "configs" / "models"
CONFIGS = [
    CONFIG_DIR / "exp26_coarse_dropout_5fold" / "resnet18.yaml",
    CONFIG_DIR / "exp32_efficientnet_b0_5fold" / "efficientnet_b0.yaml",
    CONFIG_DIR / "exp33_regnety_016" / "regnety_016.yaml",
    CONFIG_DIR / "exp34_convnext_tiny" / "convnext_tiny.yaml",
]
MARGIN = 0.05  # plateau: F1 >= max - MARGIN (exp37, exp38)
PATH = SUBMISSIONS / "ensemble-4-mean__plateau.csv"
```

## Predict


```python
submission = MeanEnsemble(CONFIGS, PATH, margin=MARGIN)
path = submission.write()
pd.read_csv(path).iloc[:, 1:].mean().round(3)  # predicted positive rate per class
```

## Upload


```python
submission.submit(
    "exp38 mean ensemble of 4 models (ResNet-18, EfficientNet-B0, RegNetY-016, ConvNeXt-Tiny) with Coarse Dropout on non VIP, 5-fold, OOF plateau thresholds",
)
```
