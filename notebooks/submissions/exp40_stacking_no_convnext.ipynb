# Submission: exp40 stacking without ConvNeXt-Tiny, C = 0.1, normalized weights, plateau thresholds

Card: [exp40_stacking_no_convnext.md](../../docs/experiments/exp40_stacking_no_convnext.md)


```python
import pandas as pd

from core.training.config import ROOT, SUBMISSIONS

from core.training.ensemble import Stacking

CONFIG_DIR = ROOT / "configs" / "models"
CONFIGS = [
    CONFIG_DIR / "exp26_coarse_dropout_5fold" / "resnet18.yaml",
    CONFIG_DIR / "exp32_efficientnet_b0_5fold" / "efficientnet_b0.yaml",
    CONFIG_DIR / "exp33_regnety_016" / "regnety_016.yaml",
]
C = 0.1  # L2 strength, exp39 conclusion
NORMALIZE = True  # logit divided by sum of member weights per class
MARGIN = 0.05  # plateau: F1 >= max - MARGIN (exp37, exp38)
PATH = SUBMISSIONS / "ensemble-3-stacking-c01-normalized__plateau.csv"
```

## Predict


```python
submission = Stacking(CONFIGS, PATH, C, margin=MARGIN, normalize=NORMALIZE)
path = submission.write()
pd.read_csv(path).iloc[:, 1:].mean().round(3)  # predicted positive rate per class
```

## Upload


```python
submission.submit(
    "exp40 logistic regression stacking C=0.1, weights normalized per class, 3 models (exp26, EfficientNet-B0, RegNetY-016), 5-fold, OOF plateau thresholds",
)
```
