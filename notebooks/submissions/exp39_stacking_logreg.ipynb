# Submission: exp39 logistic regression stacking C = 0.1, plateau thresholds

Card: [exp39_stacking_logreg.md](../../docs/experiments/exp39_stacking_logreg.md)


```python
import pandas as pd

from core.training.config import ROOT, SUBMISSIONS

from core.training.ensemble import Stacking

CONFIG_DIR = ROOT / "configs" / "models"
CONFIGS = [
    CONFIG_DIR / "exp26_coarse_dropout_5fold" / "resnet18.yaml",
    CONFIG_DIR / "exp32_efficientnet_b0_5fold" / "efficientnet_b0.yaml",
    CONFIG_DIR / "exp33_regnety_016" / "regnety_016.yaml",
    CONFIG_DIR / "exp34_convnext_tiny" / "convnext_tiny.yaml",
]
C = 0.1  # L2 strength, exp39 conclusion
MARGIN = 0.05  # plateau: F1 >= max - MARGIN (exp37, exp38)
PATH = SUBMISSIONS / "ensemble-4-stacking-c01__plateau.csv"
```

## Predict


```python
submission = Stacking(CONFIGS, PATH, C, margin=MARGIN)
path = submission.write()
pd.read_csv(path).iloc[:, 1:].mean().round(3)  # predicted positive rate per class
```

## Upload


```python
submission.submit(
    "exp39 logistic regression stacking C=0.1 of 4 models (exp26, EfficientNet-B0, RegNetY-016, ConvNeXt-Tiny), 5-fold, OOF plateau thresholds",
)
```
