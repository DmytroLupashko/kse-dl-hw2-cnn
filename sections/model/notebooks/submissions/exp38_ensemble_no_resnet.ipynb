# Submission: exp38 mean ensemble without ResNet-18, plateau thresholds

Card: [exp38_ensemble_plateau_thresholds.md](../../experiments/exp38_ensemble_plateau_thresholds.md)

```python
import os
import tempfile
from pathlib import Path

import pandas as pd

from core.config import ROOT, SUBMISSIONS
from core.cv import Config
from core.data import read_test
from core.submission import make_ensemble_submission, submit

SMOKE = os.environ.get("SMOKE") == "1"
CONFIG_DIR = ROOT / "sections" / "model" / "configs"
CONFIGS = [
    Config.from_toml(CONFIG_DIR / "exp32_efficientnet_b0_5fold" / "efficientnet_b0.toml"),
    Config.from_toml(CONFIG_DIR / "exp33_regnety_016" / "regnety_016.toml"),
    Config.from_toml(CONFIG_DIR / "exp34_convnext_tiny" / "convnext_tiny.toml"),
]
MARGIN = 0.05  # plateau: F1 >= max - MARGIN (exp37, exp38)
if SMOKE:
    TEST = read_test().head(16)
    PATH = Path(tempfile.mkdtemp()) / "smoke.csv"
else:
    TEST = None
    PATH = SUBMISSIONS / "ensemble-3-mean-no-resnet__plateau.csv"
```

## Predict

```python
path = make_ensemble_submission(CONFIGS, PATH, test=TEST, margin=MARGIN)
pd.read_csv(path).iloc[:, 1:].mean().round(3)  # predicted positive rate per class
```

## Upload

```python
if not SMOKE:
    submit(path, "exp38 mean ensemble of 3 models without ResNet-18 (EfficientNet-B0, RegNetY-016, ConvNeXt-Tiny), 5-fold, OOF plateau thresholds")
```
