# Submission: exp39 logistic regression stacking C = 0.1, plateau thresholds

Card: [exp39_stacking_logreg.md](../../experiments/exp39_stacking_logreg.md)


```python
import os
import tempfile
from pathlib import Path

import pandas as pd

from core.config import ROOT, SUBMISSIONS
from core.cv import Config
from core.data import read_test
from core.submission import Stacking

SMOKE = os.environ.get("SMOKE") == "1"
CONFIG_DIR = ROOT / "sections" / "model" / "configs"
CONFIGS = [
    Config.from_toml(CONFIG_DIR / "exp26_coarse_dropout_5fold" / "resnet18.toml"),
    Config.from_toml(
        CONFIG_DIR / "exp32_efficientnet_b0_5fold" / "efficientnet_b0.toml"
    ),
    Config.from_toml(CONFIG_DIR / "exp33_regnety_016" / "regnety_016.toml"),
    Config.from_toml(CONFIG_DIR / "exp34_convnext_tiny" / "convnext_tiny.toml"),
]
C = 0.1  # L2 strength, exp39 conclusion
MARGIN = 0.05  # plateau: F1 >= max - MARGIN (exp37, exp38)
if SMOKE:
    TEST = read_test().head(16)
    PATH = Path(tempfile.mkdtemp()) / "smoke.csv"
else:
    TEST = None
    PATH = SUBMISSIONS / "ensemble-4-stacking-c01__plateau.csv"
```

## Predict


```python
submission = Stacking(CONFIGS, PATH, C, margin=MARGIN)
path = submission.write(TEST)
pd.read_csv(path).iloc[:, 1:].mean().round(3)  # predicted positive rate per class
```

## Upload


```python
if not SMOKE:
    submission.submit(
        "exp39 logistic regression stacking C=0.1 of 4 models (exp26, EfficientNet-B0, RegNetY-016, ConvNeXt-Tiny), 5-fold, OOF plateau thresholds",
    )
```
