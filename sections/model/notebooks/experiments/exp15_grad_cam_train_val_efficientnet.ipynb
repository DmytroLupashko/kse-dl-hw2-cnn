```python
import os

import matplotlib.pyplot as plt

from core.config import ROOT
from core.cv import VIP, Config
from core.gradcam import GradCamFigures

SMOKE = os.environ.get("SMOKE") == "1"
if SMOKE:
    N_IMAGES = 2  # per figure
else:
    N_IMAGES = None  # all labeled positives
CONFIG = Config.from_toml(
    ROOT
    / "sections"
    / "model"
    / "configs"
    / "exp10_efficientnet"
    / "efficientnet_b0.toml"
)
FIGURES = ROOT / "sections" / "model" / "figures"
PREFIX = "exp15_grad_cam_train_val_efficientnet"
SAVE_FIGURES = not SMOKE

plt.rcParams.update({"figure.facecolor": "#fcfcfb", "savefig.facecolor": "#fcfcfb"})
```

## Fold models


```python
cams = GradCamFigures(CONFIG)
```

## Validation


```python
for cls in VIP:
    fig = cams.positives(cls, train=False, n=N_IMAGES)
    if SAVE_FIGURES:
        fig.savefig(FIGURES / f"{PREFIX}_{cls.lower()}_validation.png", dpi=100)
    plt.show()
```

## Train


```python
for cls in VIP:
    fig = cams.positives(cls, train=True, n=N_IMAGES)
    if SAVE_FIGURES:
        fig.savefig(FIGURES / f"{PREFIX}_{cls.lower()}_train.png", dpi=100)
    plt.show()
```
